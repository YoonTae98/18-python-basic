```python
# 집합 연산 사용하기
## 합집합
## 세트1 | 세트2
## set.union(세트1, 세트2) => 세트1.union(세트2)

a = {1, 2, 3, 4}
b = {3, 4, 5, 6}
print(a | b)
print(set.union(a, b))
print(a.union(b))
```

```python
# 교집합
## 세트1 & 세트2
## set.intersection(a, b)
##a.intersection(b)

a = {1, 2, 3, 4}
b = {3, 4, 5, 6}
print(a & b)
print(set.intersection(a, b))
print(a.intersection(b))
```

```python
## a - b
## set.difference(a, b)
## a.difference(b)

a = {1, 2, 3, 4}
b = {3, 4, 5, 6}
print(a - b)
print(set.difference(a, b))
print(a.difference(b))
print("=" * 20)
print(b - a)
```

```python
# 대칭차집합
## 세트1 ^ 세트2
## set.symmetiric_difference(a, b)
print(a ^ b)
print(set.symmetric_difference(a, b))
```

```python
# 집합 연산 후 할당 연산자 사용하기
## 세트1 |= 세트2
## 세트1.update(세트2)

a = {1, 2, 3, 4}
a |= {5}
print(a)

a = {1, 2, 3, 4}
a.update({5})
print(a)
```

```python
# intersection_update
## 세트1 &= 세트2
## 세트1.intersection_update(세트2)

a = {1, 2, 3, 4}
a &= {0, 1, 2, 3, 4}
print(a)

a = {1, 2, 3, 4}
a.intersection_update({0, 1, 2, 3, 4})
print(a)
```

```python
a = {1, 2, 3, 4}
a ^= {3, 4, 5, 6}
print(a)
```

```python
# unit 26. 심사문제

first, second = map(int,input().split())

a = {i for i in range(1, first + 1) if first % i == 0}
b = {i for i in range(1, second + 1) if second % i == 0}

divisor = a & b

result = 0
if type(divisor) == set:
    result = sum(divisor)

print(result)
```
