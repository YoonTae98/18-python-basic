```python
result = 0
def add(num):
    global result
    result += num
    return result

print(add(5))
print(add(10))
print(add(3))
```

```python
result2 = 0
def add2(num):
    global result2
    result2 += num
    return result2

print(add2(3))
print(add2(5))
```

```python
class Calculator():
    def __init__(self):
        self.result = 0

    def add(self, num):
        self.result += num
        return self.result

cal1 = Calculator()
print(cal1.add(5))
print(cal1.add(10))

cal2 = Calculator()
print(cal2.add(3))
print(cal2.add(20))
```

```python
# 속성 사용하기
# class 클래스이름:
#     def __init__(self):
#         self.속성 = 값

class Person:
    def __init__(self):
        self.hello = "안녕하세요"

    def greeting(self):
        print(self.hello)

james = Person()
print(james.hello)
james.greeting()
```

```python
# 인스턴스를 만들 때 값 받기

# class 클래스이름:
#     def __init__(self, 매개변수1, 매개변수2):
#         self.속성1 = 매개변수1
#         self.속성2 = 매개변수2

class Person:
    def __init__(self, name, age, address):
        self.hello = "안녕하세요."
        self.name = name
        self.age = age
        self.address = address

    def greeting(self):
        print(f"{self.hello} 저는 {self.name}입니다.")

maria = Person("마리아", 28, "서울시 서초구 반포동")
maria.greeting()

print('이름', maria.name)
print('나이', maria.age)
print('주소', maria.address)
```

```python
# unit 34. 연습문제

class Knight:
    def __init__(self, health, mana, armor):
        self.health = health
        self.mana = mana
        self.armor = armor

    def slash(self):
        print("베기")

x = Knight(health=542.4, mana=210.3, armor=38)
print(x.health, x.mana, x.armor)
x.slash()
```

```python
# unit 34. 심사문제
# 511.68 334.0 298
# 1803.68 1184.0 645

class Annie:
    def __init__(self, health, mana, ability_power):
        self.health = health
        self.mana = mana
        self.ability_power = ability_power

    def tibbers(self):
        print(f"티버: 피해량 {self.ability_power * 0.65 + 400}")

health, mana, ability_power = map(float, input().split())

x = Annie(health=health, mana=mana, ability_power=ability_power)
x.tibbers()
```
