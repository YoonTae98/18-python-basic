```python
# while 반복문 사용하기

# 초기식
# while 조건식:
#     반복할코드
#     변화식

i = 0
while i < 10:
    print("hello world")
    i += 1 # i = i + 1
```

```python
# 초기값을 1부터 시작하기
i = 1
while i <= 5:
    print("hello world", i)
    i += 1
```

```python
# 초기값을 감소시키기
i = 5
while i > 0:
    print("hello", i)
    i -= 1 # i = i - 1
```

```python
count = int(input())

i = 0
while i < count:
    print("hello", i)
    i += 1
```

```python
# 연습문제

i = 2
j = 5

while i <= 32 or j >= 1:
    print(i, j)
    i *= 2
    j -= 1
```

```python
# 심사문제
금액 = int(input())
요금 = 1350

while 금액 >= 요금:
    금액 -= 요금
    print(금액)
```
