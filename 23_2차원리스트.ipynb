```python
# 반복문으로 2차원 리스트의 요소를 모두 출력하기
a = [
    [10, 20],
    [30, 40],
    [50, 60]
]

for x, y in a:
    print(x, y, end=' ')
```

```python
a = [
    [10, 20],
    [30, 40],
    [50, 60]
]
for i in a:
    for j in i:
        print(j, end = ' ')
    print()
```

```python
a = [
    [10, 20],
    [30, 40],
    [50, 60]
]
for i in range(len(a)):
    for j in range(len(a[i])):
        print(a[i][j], end = ' ')
    print()
```

```python
a = [
    [10, 20],
    [30, 40],
    [50, 60]
]
i = 0
while i < len(a):
    x, y = a[i]
    print(x, y)
    i += 1
```

```python
# while 반복문을 두 번 사용하기
a = [
    [10, 20],
    [30, 40],
    [50, 60]
]
i = 0
while i < len(a):
    j = 0
    while j < len(a[i]):
        print(a[i][j], end = ' ')
        j += 1
    print()
    i += 1
```

```python
# for 반복문으로 1차원 리스트 만들기
a = []

for _ in range(10):
    a.append(0)

print(a)
```

```python
# for 반복문으로 2차원 리스트 만들기
a = []
for i in range(3):
    line = []
    for j in range(2):
        line.append(0)
    a.append(line)
print(a)
```

```python
# unit 23. 심사문제

col, row = map(int, input().split())

matrix = []
for i in range(row):
    matrix.append(list(input()))

for i in range(row):
    result = ""
    for j in range(col):
        if matrix[i][j] == "*":
            result += "*"
        else:
            지뢰수 = 0
            for x in range(-1, 2):
                for y in range(-1, 2):
                    find1 = i + x
                    find2 = j + y
                    if 0 <= find1 < row and 0 <= find2 < col:
                        if matrix[find1][find2] == "*":
                            지뢰수 += 1
            result += str(지뢰수)
    print(result)
```
