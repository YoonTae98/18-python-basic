```python
가격 = int(input())
쿠폰이름 = input()

if 쿠폰이름[:4] == "Cash":
    할인 = int(쿠폰이름[4:])
    print(가격 - 할인)
```

```python
국어, 영어, 수학, 과학 = map(int, input().split())
평균 = (국어 + 영어 + 수학 + 과학) / 4
if 평균 >= 80:
    print("합격")
else:
    print("불합격")
```

```python
age = int(input())
balace = 9000
if 12 >= age >= 7:
    print(balace - 650)
elif 18 >= age:
    print(balace - 1050)
else:
    print(balace - 1250)
```

```python
a = int(input())
for i in range(0, 10, 1):
    result = a * i
    print(f"{a} * {i} = {result}")
```

```python
금액 = int(input())
while 금액 >= 1350:
    print(금액 - 1350)
    금액 -= 1350
```

```python
start, stop = map(int, input().split())

i = start

while True:
    if i > stop or i < 1 or i > 200 or stop < 10 or stop > 200:
        break
    elif i % 10 == 3:
        i += 1
        continue
    print(i, end=' ')
    i += 1
```
