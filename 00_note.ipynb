```python
# 커피숍 계산 프로그램

재고 = 100
가격 = 1500
원가 = 700

일매출 = 0
일수익 = 0

월매출 = 0
월수익 = 0


# 커피 주문 및 계산 함수
def coffee(cash, cup):
    global 재고, 일매출, 일수익, 월매출, 월수익

    cash = int(cash)
    cup = int(cup)

    # 재고 확인
    if cup > 재고:
        print("재고가 부족합니다.")
        return

    # 결제금액 확인
    총금액 = 가격 * cup

    if cash < 총금액:
        print("지불금액이 부족합니다.")
        return

    # 계산
    재고 -= cup
    잔돈 = cash - 총금액

    # 매출과 수익 계산
    매출 = 총금액
    수익 = (가격 - 원가) * cup

    일매출 += 매출
    일수익 += 수익

    월매출 += 매출
    월수익 += 수익

    print("주문 수량:", cup, "잔")
    print("결제 금액:", 총금액, "원")
    print("잔돈:", 잔돈, "원")
    print("남은 재고:", 재고, "잔")

    return 잔돈


# 일별 매출과 수익을 보여주는 함수
def daily_sales():
    print("\n===== 일별 매출 현황 =====")
    print("일 매출:", 일매출, "원")
    print("일 수익:", 일수익, "원")


# 월별 매출과 수익을 보여주는 함수
def monthly_sales():
    print("\n===== 월별 매출 현황 =====")
    print("월 매출:", 월매출, "원")
    print("월 수익:", 월수익, "원")


# 재고 확인 함수
def stock():
    print("\n===== 재고 현황 =====")
    print("남은 커피:", 재고, "잔")


coffee(5000, 2)
coffee(10000, 3)

daily_sales()
monthly_sales()
stock()
```

```python
def analyze_scores(*args):
    average = sum(args) / len(args)
    count = 0

    for score in args:
        if score >= average:
            count += 1

    print(f"평균: {average}")
    print(f"평균 이상인 수의 개수: {count}")


scores = map(int, input().split())
analyze_scores(*scores)
```

```python
## 9
# 70 80 90 100 50
def analyze_scores(*args):
    average = sum(args) / len(args)

    count = 0

    for score in args:
        if score >= average:
            count += 1

    print(f"평균: {average}", end="")
    print(f"평균 이상인 수의 개수: {count}")


# 입력
scores = list(map(int, input().split()))

analyze_scores(*scores)
```

```python
## 10
inventory = {
    "노트북": 1500000,
    "마우스": 30000,
    "키보드": 50000,
    "모니터": 300000
}


def checkout(**kwargs):
    total = 0
    for product, quantity in kwargs.items():
        if product in inventory:
            price = inventory[product] * quantity
            print(f"{product} {quantity}개: {price:,}원")
            total += price
        else:
            print(f"판매하지 않는 상품입니다: {product}")

    print(f"총 결제 금액: {total:,}원")
    print()


checkout(노트북=1, 마우스=2)
checkout(모니터=2, 패드=1)
```

```python
# 1. Product 클래스
class Product:
    """상품 정보를 관리하는 클래스"""

    def __init__(self, product_id, name, price, weight, stock):
        self.product_id = product_id
        self.name = name
        self.price = price
        self.weight = weight
        self.stock = stock

    def get_info(self):
        """상품 정보를 딕셔너리로 반환"""
        return {
            "product_id": self.product_id,
            "name": self.name,
            "price": self.price,
            "weight": self.weight,
            "stock": self.stock
        }

    def update_stock(self, quantity):
        """
        재고 수량 업데이트
        재고가 부족하면 False 반환, 성공하면 True 반환
        """
        if self.stock < quantity:
            return False

        self.stock -= quantity
        return True

    def apply_discount(self, discount_rate):
        """할인율을 적용한 가격 반환 (원본 가격은 유지)"""
        return self.price * (1 - discount_rate / 100)


# 2. ShoppingCart 클래스
class ShoppingCart:
    """장바구니를 관리하는 클래스"""

    def __init__(self):
        """장바구니 초기화"""
        self.products = []

    def add_product(self, product, quantity):
        """
        장바구니에 상품 추가
        {'product': Product객체, 'quantity': 수량} 형태로 저장
        """

        # 재고 확인
        if product.update_stock(quantity):
            self.products.append({
                "product": product,
                "quantity": quantity
            })
            print(f"{product.name} {quantity}개가 장바구니에 추가되었습니다.")
        else:
            print(f"{product.name}의 재고가 부족합니다.")

    def remove_product(self, product_id):
        """상품 ID로 장바구니에서 상품 제거"""

        for item in self.products:
            if item["product"].product_id == product_id:
                self.products.remove(item)
                print("상품이 장바구니에서 제거되었습니다.")
                return

        print("해당 상품이 장바구니에 없습니다.")

    def get_total_price(self):
        """장바구니 내 모든 상품의 총 금액 계산"""

        total = 0

        for item in self.products:
            product = item["product"]
            quantity = item["quantity"]

            total += product.price * quantity

        return total

    def get_total_weight(self):
        """장바구니 내 모든 상품의 총 무게 계산"""

        total = 0

        for item in self.products:
            product = item["product"]
            quantity = item["quantity"]
            
            total += product.weight * quantity

        return total

    def show_cart(self):
        """
        장바구니 내용을 보기 좋게 출력
        """

        print("\n===== 장바구니 =====")

        for item in self.products:
            product = item["product"]
            quantity = item["quantity"]

            print(
                f"상품명: {product.name} | "
                f"수량: {quantity}개 | "
                f"가격: {product.price * quantity:,}원"
            )

        print("-------------------")
        print(f"총 금액: {self.get_total_price():,}원")
        print(f"총 무게: {self.get_total_weight():.1f}kg")


# 3. Order 클래스
class Order:
    """주문 정보를 관리하는 클래스"""

    def __init__(self, order_id, cart, customer_name):
        self.order_id = order_id
        self.cart = cart
        self.customer_name = customer_name
        self.shipping_fee = 3000

    def calculate_final_price(self):
        """
        최종 결제 금액 계산
        상품 금액 + 배송비
        """

        return self.cart.get_total_price() + self.shipping_fee

    def print_receipt(self):
        """
        주문서 출력
        """

        print("\n===== 주문서 =====")
        print(f"주문 ID: {self.order_id}")
        print(f"고객명: {self.customer_name}")

        print("\n[주문 상품]")

        for item in self.cart.products:
            product = item["product"]
            quantity = item["quantity"]

            print(
                f"{product.name} "
                f"x {quantity}개 "
                f"= {product.price * quantity:,}원"
            )

        product_total = self.cart.get_total_price()

        print("\n-------------------")
        print(f"상품 금액: {product_total:,}원")
        print(f"배송비: {self.shipping_fee:,}원")
        print(f"최종 결제 금액: {self.calculate_final_price():,}원")
        print("===================")


# 실행예시 코드

# 상품 생성
laptop = Product("P001", "노트북", 1200000, 2.5, 10)
mouse = Product("P002", "무선마우스", 35000, 0.2, 50)
keyboard = Product("P003", "기계식키보드", 89000, 1.0, 30)

# 장바구니 생성 및 상품 추가
cart = ShoppingCart()
cart.add_product(laptop, 1)
cart.add_product(mouse, 2)
cart.add_product(keyboard, 1)

# 장바구니 확인
cart.show_cart()

# 주문 생성
order = Order("ORD20260211001", cart, "홍길동")

# 주문서 출력
order.print_receipt()
```
