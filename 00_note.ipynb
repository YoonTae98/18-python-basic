```python
# 2. 당신은 커피숍 계산 프로그램을 함수로 만들려고 한다. 
# 커피가 주문되고 계산이 이루어지는 함수와 일별, 월별 매출과 수익을 보여주는 함수를 자유롭게 만들어주세요

def coffee(cash, cup):
    재고 = 100
    가격 = 1500
    if 가격 * int(cup) < 재고 * 가격:
        print("재고가 부족합니다.")
    elif int(cash) < 가격:
        print("지불금액이 부족합니다.")
    elif 가격 * int(cup) >= 재고 * 가격:
        재고 -= int(cup)
        잔돈 = int(cash) - int(cup) * 가격
        return
```

```python
# 커피숍 계산 프로그램

재고 = 100
가격 = 1500
원가 = 700

일매출 = 0
일수익 = 0

월매출 = 0
월수익 = 0


# 커피 주문 및 계산 함수
def coffee(cash, cup):
    global 재고, 일매출, 일수익, 월매출, 월수익

    cash = int(cash)
    cup = int(cup)

    # 재고 확인
    if cup > 재고:
        print("재고가 부족합니다.")
        return

    # 결제금액 확인
    총금액 = 가격 * cup

    if cash < 총금액:
        print("지불금액이 부족합니다.")
        return

    # 계산
    재고 -= cup
    잔돈 = cash - 총금액

    # 매출과 수익 계산
    매출 = 총금액
    수익 = (가격 - 원가) * cup

    일매출 += 매출
    일수익 += 수익

    월매출 += 매출
    월수익 += 수익

    print("주문 수량:", cup, "잔")
    print("결제 금액:", 총금액, "원")
    print("잔돈:", 잔돈, "원")
    print("남은 재고:", 재고, "잔")

    return 잔돈


# 일별 매출과 수익을 보여주는 함수
def daily_sales():
    print("\n===== 일별 매출 현황 =====")
    print("일 매출:", 일매출, "원")
    print("일 수익:", 일수익, "원")


# 월별 매출과 수익을 보여주는 함수
def monthly_sales():
    print("\n===== 월별 매출 현황 =====")
    print("월 매출:", 월매출, "원")
    print("월 수익:", 월수익, "원")


# 재고 확인 함수
def stock():
    print("\n===== 재고 현황 =====")
    print("남은 커피:", 재고, "잔")


coffee(5000, 2)
coffee(10000, 3)

daily_sales()
monthly_sales()
stock()
```
