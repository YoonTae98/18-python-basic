```python
# 지뢰찾기

col, row = map(int, input().split())

matrix = []
for i in range(row):
    matrix.append(list(input()))
```

```python
print(matrix)
```

```python
for i in range(row):
    result = ""
    for j in range(col):
        if matrix[i][j] == "*":
            result += "*"
        else:
            지뢰수 = 0
            for x in range(-1, 2):
                for y in range(-1, 2):
                    find1 = x + i
                    find2 = y + j
                    if 0 <= find1 < row and 0 <= find2 < col:
                        if matrix[find1][find2] == "*":
                            지뢰수 += 1
            result += str(지뢰수)
    print(result)
```
